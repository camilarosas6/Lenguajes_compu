## Datos iniciales

Se tiene el alfabeto:

\[
\Sigma = \{a,b\}

\[
L = \{\lambda,a,b\}
\]

\[
M = \{b,aa\}

```python
import requests
```

## 1. Unión de lenguajes

Se tienen los lenguajes:

\[
L = \{\lambda, a, b\}
\]

\[
M = \{b, aa\}
\]

La operación solicitada es:

\[
L \cup M
\]

Para obtener el resultado se utiliza el siguiente endpoint:

`POST /lenguajes/union`

```python
url = "http://127.0.0.1:8000/lenguajes/union"

datos = {
    "lenguaje1": ["", "a", "b"],
    "lenguaje2": ["b", "aa"]
}

respuesta = requests.post(url, json=datos)

respuesta.json()
```

### Resultado

El endpoint devuelve:

\[
L \cup M = \{\lambda, a, b, aa\}
\]

## 2. Intersección de lenguajes

Se calcula:

## 6. Concatenación de lenguajes

Se calcula:

\[

\]

utilizando el endpoint:

`POST /lenguajes/concatenar`

Para obtener \(ML\), se coloca \(M\) como primer lenguaje

## 10. \((LM)\cup(M^*L^2)\)

Para resolver este ejercicio se descompone la expresión en varias
operaciones.

Primero se obtiene \(L^2\) mediante el endpoint de potencia de
lenguajes.

Luego se obtiene \(M^2\)

```python
url = "http://127.0.0.1:8000/lenguajes/potencia"

datos = {
    "lenguaje": ["", "a", "b"],
    "exponente": 2
}

respuesta = requests.post(url, json=datos)

L2 = set(respuesta.json()["resultado"])

L2
```

```python
url = "http://127.0.0.1:8000/lenguajes/kleene"

datos = {
    "lenguaje": ["b", "aa"],
    "max_elementos": 8
}

respuesta = requests.post(url, json=datos)

M_kleene = set(respuesta.json()["resultado"])

M_kleene
```

```python
url = "http://127.0.0.1:8000/lenguajes/concatenar"

datos = {
    "lenguaje1": list(M_kleene),
    "lenguaje2": list(L2)
}

respuesta = requests.post(url, json=datos)

M_kleene_L2 = set(respuesta.json()["resultado"])

len(M_kleene_L2)
```

### Visualización de cadenas

```python
sorted(M_kleene_L2, key=lambda x: (len(x), x))
```

### Unión con POST /lenguajes/union

```python
url = "http://127.0.0.1:8000/lenguajes/concatenar"

datos = {
    "lenguaje1": ["", "a", "b"],
    "lenguaje2": ["b", "aa"]
}

respuesta = requests.post(url, json=datos)

LM = set(respuesta.json()["resultado"])
```

```python
url = "http://127.0.0.1:8000/lenguajes/union"

datos = {
    "lenguaje1": list(LM),
    "lenguaje2": list(M_kleene_L2)
}

respuesta = requests.post(url, json=datos)

resultado_final = respuesta.json()["resultado"]

sorted(resultado_final, key=lambda x: (len(x), x))
```

### Resultado final

Después de calcular \(LM\), \(L^2\) y los primeros 8 elementos de
\(M^*\), se calculó \(M^*L^2\) mediante el endpoint de concatenación
de lenguajes.

Finalmente, se utilizó el endpoint de unión para obtener:

\[
(LM)\cup(M^*L^2)
\]
